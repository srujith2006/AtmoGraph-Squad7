```python
import pandas as pd

data = {
    "Supplier": ["ABC Electronics", "XYZ Chips", "Global Parts", "Tech Components"],
     "Factory": ["Factory A", "Factory B", "Factory C", "Factory D"],
     "Port": ["Shanghai Port", "Mumbai port", "Singapore Port", "Delhi Hub"],
     "Risk": ["High","Medium","Low","medium"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Check dataset information
print("Dataset Information:")
print(df.info())

# Count total suppliers
print("Total Suppliers:", len(df))

# Count Risk Levels
print("\nRisk Level Count:")
print(df["Risk"].value_counts())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
import networkx as nx

print("NetworkX Imported Successfully!")
```

```python
G = nx.DiGraph()

for index, row in df.iterrows():
    G.add_edge(row["Supplier"], row["Factory"])
    G.add_edge(row["Factory"], row["Port"])

print("Total Nodes:", G.number_of_nodes())
print("Total Edges:", G.number_of_edges())

print("\nAll Nodes:")
print(list(G.nodes()))

print("\nAll Connections:")
print(list(G.edges()))
```

```python
import matplotlib.pyplot as plt

print("Matplotlib Imported Successfully!")
```

```python
plt.figure(figsize=(12, 7))

pos = nx.spring_layout(G, seed=42)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="lightblue",
    node_size=2500,
    font_size=8,
    arrows=True
)

plt.title("AtmoGraph - Supply Chain Network")
plt.show()
```

```python
# Add Risk Information to Supplier Nodes

for index, row in df.iterrows():
    G.nodes[row["Supplier"]]["Risk"] = row["Risk"]

print("Supplier Risk Analysis:")

for supplier in df["Supplier"]:
    print(supplier, ":", G.nodes[supplier]["Risk"])
```

```python
print("Total Suppliers:", len(df))

print("Total Nodes:", G.number_of_nodes())

print("Total Edges:", G.number_of_edges())

print("Is Graph Directed:", nx.is_directed(G))

print("Graph Testing Completed Successfully!")
```
